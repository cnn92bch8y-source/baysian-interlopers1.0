```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Plot (3,5,10), Plot the projections on the xy, xz, and yz planes

#target points
px=3
py=5
pz=10
pv=13

#blank graph
fig=plt.figure()
ax=fig.add_subplot(111,projection='3d') #3d cords
ax.scatter(px,py,pz,color='magenta',marker='o',s=100,label='Target Point') #plot og plot

#plot projections on plane
ax.scatter(px,py,0,color='orange',marker='o',s=100,label='Projection on XY Plane') #set z=0
ax.scatter(px,0,pz,color='black',marker='o',s=100,label='Projection on XZ Plane') #set y=0
ax.scatter(0,py,pz,color='purple',marker='o',s=100,label='Projection on YZ Plane') #set x=0
ax.scatter(px,py,pv,color='green',marker='o',s=100,label='Projection on Sphere') #set x=0

#connect lines 
ax.plot3D([px,px],[py,py],[0,pz],color='orange',linestyle='--') #line to xy plane
ax.plot3D([px,px],[0,py],[pz,pz],color='black',linestyle='dashed') #line to xz plane
ax.plot3D([0,px],[py,py],[pz,pz],color='purple',linestyle='dashed') #line to yz plane 
ax.plot3D([px,px],[py,py],[pz,pv],color='green',linestyle='dashed') #line to sphere

ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_zlabel('Z-axis')
ax.set_title('3D Point and its Projections on Coordinate Planes')
ax.legend()
plt.show()
```

```python
#points prof. lee gave me to us
px=3
py=5
pz=10

#3d graph window 
fig= plt.figure() 
ax=fig.add_subplot(111, projection='3d')
ax.scatter(px,py,pz,color='pink', s=100, label='OG Point')
ax.set_title('1st 3D Point and Projections')
#lpane projections, xy plane set z=0
ax.scatter(px, py, 0, color='teal', s=50, label='XY Projection')

#xz plane set y=0
ax.scatter(px, 0, pz, color='purple', s=50, label='XZ Projection')

#yz plane set x=0
ax.scatter(0, py, pz, color='magenta', s=50, label='YZ Projection')

#ax.plot([start_x, end_x], [start_y, end_y], [start_z, end_z])
#only z changes from pz to 0
ax.plot([px,px],[py,py], [pz,0], color='teal', linestyle='dashed')
#only y changes from py to 0
ax.plot([px,px],[py,0], [pz,pz], color='purple', linestyle='dashed')
#only x changes from px to 0
ax.plot([px,0],[py,py], [pz,pz], color='magenta', linestyle='dashed')


# Generate random numbers for the normal vector coordinates
random_array = np.random.randn(3)
A = random_array[0]
B = random_array[1]
C = random_array[2]

# Calculate length manually to normalize the vector to a unit vector
vec_length = np.sqrt(A**2 + B**2 + C**2)
A = A / vec_length
B = B / vec_length
C = C / vec_length

# Set static limits for the axes so the view is clean
ax.set_xlim([-15, 15])
ax.set_ylim([-15, 15])
ax.set_zlim([-15, 15])

# Make a simple grid to draw the plane surface
x_grid = np.linspace(-15, 15, 10)
y_grid = np.linspace(-15, 15, 10)
X, Y = np.meshgrid(x_grid, y_grid)

# Solve for Z using the plane equation: Ax + By + Cz = 0
Z = (-A * X - B * Y) / C
ax.plot_surface(X, Y, Z, color='cyan', alpha=0.3)

#project point onto plane
# Find distance using the dot product formula manually
distance = (px * A) + (py * B) + (pz * C)

# Calculate the coordinates for the new point on the plane LEARN HOW TO FLUIDLY EXPLAIN THIS :
proj_x = px - (distance * A)
proj_y = py - (distance * B)
proj_z = pz - (distance * C)

# Plot the projection point on the plane and draw the connecting line
# Plot the new projection point as a magenta star marker ('*')
ax.scatter(proj_x, proj_y, proj_z, color='orange', marker='s', s=120, label='Plane Projection')
ax.plot([px, proj_x], [py, proj_y], [pz, proj_z], color='orange', linewidth=2)


# --- STEP 5: LABELS AND SHOW ---
ax.set_xlabel('X Axis')

ax.set_ylabel('Y Axis')
ax.set_zlabel('Z Axis')
ax.set_title('3D Vector Projections Graph')
ax.legend()

plt.show()
```

```python
import math
import random

# Generate a mock 3D point to project
point_3d = [1.0, 2.0, 3.0]

#take answers u get 

# NON-UNIFORM DISTRIBUTION (Euler Angles)
# Separately picking 3 random angles causes clustering at the poles.
yaw = random.uniform(0, 2 * math.pi)
pitch = random.uniform(0, 2 * math.pi)
roll = random.uniform(0, 2 * math.pi)

x, y, z = point_3d

# Rotate around Z-axis (Yaw)
x1 = x * math.cos(yaw) - y * math.sin(yaw)
y1 = x * math.sin(yaw) + y * math.cos(yaw)
z1 = z

# Rotate around Y-axis (Pitch)
x2 = x1 * math.cos(pitch) + z1 * math.sin(pitch)
y2 = y1
z2 = -x1 * math.sin(pitch) + z1 * math.cos(pitch)

# Rotate around X-axis (Roll)
x3 = x2
y3 = y2 * math.cos(roll) - z2 * math.sin(roll)

#look up rotation matrices and how to apply them to points in 3D space to understand this better

# Flatten onto the plane by dropping the Z coordinate
non_uniform_projection = (x3, y3)

# UNIFORM DISTRIBUTION 
# Pick a random facing direction uniformly using a normal distribution.
nx = random.gauss(0, 1)
ny = random.gauss(0, 1)
nz = random.gauss(0, 1)

# Shrink the direction vector to a length of 1
length = math.sqrt(nx**2 + ny**2 + nz**2)
nx, ny, nz =  nx / length, ny / length, nz / length

# Create a 2D grid coordinate system directly on that random plane
# Line 1 on the plane (perpendicular to the direction vector)
ux, uy, uz = -ny, nx, 0
u_len = math.sqrt(ux**2 + uy**2 + uz**2)
if u_len == 0:  # Edge case handling if pointing straight up
    ux, uy, uz = 1, 0, 0
else:
    ux, uy, uz = ux / u_len, uy / u_len, uz / u_len

# Line 2 on the plane (perpendicular to both the direction and Line 1)
wx = ny * uz - nz * uy
wy = nz * ux - nx * uz
wz = nx * uy - ny * ux

# Map the 3D point onto this uniform 2D plane grid using dot products
u_proj = point_3d[0] * ux + point_3d[1] * uy + point_3d[2] * uz
w_proj = point_3d[0] * wx + point_3d[1] * wy + point_3d[2] * wz

uniform_projection = (u_proj, w_proj)

# results
print("Non-Uniform 2D Output:", non_uniform_projection)
print("Uniform 2D Output:", uniform_projection)
```
