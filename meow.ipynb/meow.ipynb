```python

# 1. Define the function
def f(x):
    return np.sin(x**2 + 1 - x)
```

```python
# 2. Parameters
N = 10
start = 0
end = np.pi
width = (end - start) / N
```

```python
# 3. Calculate individual areas and store in a list
rectangle_areas = []

for i in range(N):
    x_i = start + i * width
    height = f(x_i)
    area = height * width
    rectangle_areas.append(area)
    
    print(f"Rectangle {i+1}: Area = {area:.4f}")
```

```python
# 4. Final approximation
total_sum = sum(rectangle_areas)

print("-" * 30)
print(f"Total Riemann Sum Approximation: {total_sum:.4f}")
```

```python
import numpy as np

# 1. Define the function
def f(x):
    return np.sin(x**2 + 1 - x)

# 2. Parameters
N = 10
start = 0
end = np.pi
width = (end - start) / N

# 3. Calculate using trapezoids
total_area = 0

print(f"{'Slice':<10} {'Left x':<10} {'Right x':<10} {'Area of Trapezoid'}")
print("-" * 60)

for i in range(N):
    x_left = start + i * width
    x_right = start + (i + 1) * width
    
    # Trapezoid area formula: ((f(left) + f(right)) / 2) * width
    height_avg = (f(x_left) + f(x_right)) / 2
    area_of_slice = height_avg * width
    
    total_area += area_of_slice
    
    print(f"{i+1:<10} {x_left:<10.3f} {x_right:<10.3f} {area_of_slice:.4f}")

print("-" * 60)
print(f"Total Trapezoidal Approximation: {total_area:.4f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Wide display setup
plt.rcParams["figure.figsize"] = (15, 5)
num_points = 10000

# Uniform Cube Surface
box_pts = []
for _ in range(num_points):
    face = np.random.randint(0, 6)
    u = np.random.uniform(-1, 1)
    v = np.random.uniform(-1, 1)
    
    if face == 0:   box_pts.append([1, u, v])
    elif face == 1: box_pts.append([-1, u, v])
    elif face == 2: box_pts.append([u, 1, v])
    elif face == 3: box_pts.append([u, -1, v])
    elif face == 4: box_pts.append([u, v, 1])
    else:           box_pts.append([u, v, -1])

box_pts = np.array(box_pts)

ax1 = plt.subplot(1, 3, 1, projection='3d')
ax1.scatter(box_pts[:, 0], box_pts[:, 1], box_pts[:, 2], c='red', s=0.1, alpha=0.5)
ax1.set_title("1. Uniform Cube Surface")
ax1.set_xlim([-1.1, 1.1])
ax1.set_ylim([-1.1, 1.1])
ax1.set_zlim([-1.1, 1.1])

# Where Vectors Hit Cube
# Generate random 3D directions
random_vectors = np.random.normal(0, 1, (num_points, 3))

# Scale them out until their dominant direction hits the boundary wall
projected_cube_pts = random_vectors / np.max(np.abs(random_vectors), axis=1, keepdims=True)

ax2 = plt.subplot(1, 3, 2, projection='3d')
ax2.scatter(projected_cube_pts[:, 0], projected_cube_pts[:, 1], projected_cube_pts[:, 2], c='red', s=0.1, alpha=0.5)
ax2.set_title("2. Where Vectors Hit Cube")
ax2.set_xlim([-1.1, 1.1])
ax2.set_ylim([-1.1, 1.1])
ax2.set_zlim([-1.1, 1.1])

# Cube Squeezed into Sphere (SIMPLIFIED)
# SIMPLIFIED: Divide each point by its Euclidean distance from center in one line
cube_to_sphere_pts = box_pts / np.linalg.norm(box_pts, axis=1, keepdims=True)

ax3 = plt.subplot(1, 3, 3, projection='3d')
ax3.scatter(cube_to_sphere_pts[:, 0], cube_to_sphere_pts[:, 1], cube_to_sphere_pts[:, 2], c='red', s=0.1, alpha=0.5)
ax3.set_title("3. Cube Squeezed into Sphere")
ax3.set_xlim([-1.1, 1.1])
ax3.set_ylim([-1.1, 1.1])
ax3.set_zlim([-1.1, 1.1])

plt.tight_layout()
plt.savefig('cube_transformations.png', dpi=150)
```

```python
# Create a list of stars with their properties
# Each star is a dictionary containing:
# mass (m), position (rx, ry, rz), and velocity (vx, vy, vz)

stars = [
    {"name": "Star Paws", "m": 1.0, "rx": 0.0, "ry": 0.0, "rz": 0.0, "vx": 0.1, "vy": 0.1, "vz": 0.0},
    {"name": "Star Tail", "m": 1.0, "rx": 2.0, "ry": 0.0, "rz": 0.0, "vx": -0.1, "vy": -0.1, "vz": 0.0},
    {"name": "Star Whiskers", "m": 0.5, "rx": 10.0, "ry": 10.0, "rz": 10.0, "vx": 5.0, "vy": -5.0, "vz": 2.0},
    {"name": "Star Ears", "m": 1.2, "rx": 2.2, "ry": 0.1, "rz": -0.1, "vx": 0.12, "vy": 0.08, "vz": -0.01}
]

# Define the Gravitational Constant. 
# In specialized astronomy code, we often set G = 1 to simplify the math.
G = 1.0 

print("Scanning the skies for binary star systems...\n")

# Use a nested loop to pair up the stars
# We use 'i' for the first star and 'j' for the second star.
# By starting 'j' at 'i + 1', we never compare a star to itself,
# and we never check the same pair twice (e.g., if we check A and B, 
# we won't waste time checking B and A later).

num_stars = len(stars)

for i in range(num_stars):
    for j in range(i + 1, num_stars):
        
        # Grab our two stars for comparison
        star1 = stars[i]
        star2 = stars[j]
        
        # Pull out the masses for easy reading
        m1 = star1["m"]
        m2 = star2["m"]
        
        # Physics calculations
        
        # Calculate the distance between the two stars using the 3D Pythagorean theorem
        dx = star2["rx"] - star1["rx"]
        dy = star2["ry"] - star1["ry"]
        dz = star2["rz"] - star1["rz"]
        distance = (dx**2 + dy**2 + dz**2)**0.5
        
        # Calculate the relative velocity components between the two stars
        dvx = star2["vx"] - star1["vx"]
        dvy = star2["vy"] - star1["vy"]
        dvz = star2["vz"] - star1["vz"]
        # Square the relative speed for the Kinetic Energy formula
        rel_speed_squared = dvx**2 + dvy**2 + dvz**2
        
        # Calculate Potential Energy (PE)
        # Formula: PE = -G * m1 * m2 / distance
        # Potential energy is negative because gravity is an attractive force.
        potential_energy = -G * m1 * m2 / distance
        
        # Calculate Kinetic Energy (KE) of the pair relative to each other
        # We use 'reduced mass' (mu) for two-body systems: mu = (m1 * m2) / (m1 + m2)
        # Formula: KE = 0.5 * mu * v^2
        reduced_mass = (m1 * m2) / (m1 + m2)
        kinetic_energy = 0.5 * reduced_mass * rel_speed_squared
        
        # Calculate Total Energy
        # Formula: Total Energy = Kinetic Energy + Potential Energy
        total_energy = kinetic_energy + potential_energy
        
        # Evaluate if Total Energy is less than 0
        # If Total Energy < 0, the stars don't have enough speed to escape 
        # each other's gravity. They are gravitationally bound (a binary pair)!
        if total_energy < 0:
            print(f"🙀 MEOW! match found: {star1['name']} and {star2['name']} form a binary system!")
            print(f"   -> Distance: {distance:.2f} units")
            print(f"   -> Total Orbital Energy: {total_energy:.4f} (Bound system)\n")
        else:
            print(f"😾 meow *sad*, {star1['name']} and {star2['name']} just passing by.")
            print(f"   -> Total Orbital Energy: {total_energy:.4f} (Unbound/Flyby)\n")

            #how can 2 binaries exist in the same system?
            #correct code. 
            #re-state old notes, negative energy means bound system, positive energy means unbound system.
            #assifgn one binary star taol n ear r deepest binary then igore the weaker ones 
            #ignore duplicates (weak ones) yikes work 
```
