```python
def f(x):
    return np.sin(x**2 + 1 - x)
```

```python
N=10
start=0
end=np.pi
x_val = np.linspace(start, end, 500)
y_val = f(x_val)
```

```python
plt.bar(x_rects, y_rects, width=width, align='edge', 
        color='red', alpha=0.3, edgecolor='black', 
        label=f'Riemann sum (N={N})')
```

```python

# 1. Define the function
def f(x):
    return np.sin(x**2 + 1 - x)
```

```python
# 2. Parameters
N = 10
start = 0
end = np.pi
width = (end - start) / N
```

```python
# 3. Calculate individual areas and store in a list
rectangle_areas = []

for i in range(N):
    x_i = start + i * width
    height = f(x_i)
    area = height * width
    rectangle_areas.append(area)
    
    print(f"Rectangle {i+1}: Area = {area:.4f}")
```

```python
# 4. Final approximation
total_sum = sum(rectangle_areas)

print("-" * 30)
print(f"Total Riemann Sum Approximation: {total_sum:.4f}")
```

```python
import numpy as np

# 1. Define the function
def f(x):
    return np.sin(x**2 + 1 - x)

# 2. Parameters
N = 10
start = 0
end = np.pi
width = (end - start) / N

# 3. Calculate using trapezoids
total_area = 0

print(f"{'Slice':<10} {'Left x':<10} {'Right x':<10} {'Area of Trapezoid'}")
print("-" * 60)

for i in range(N):
    x_left = start + i * width
    x_right = start + (i + 1) * width
    
    # Trapezoid area formula: ((f(left) + f(right)) / 2) * width
    height_avg = (f(x_left) + f(x_right)) / 2
    area_of_slice = height_avg * width
    
    total_area += area_of_slice
    
    print(f"{i+1:<10} {x_left:<10.3f} {x_right:<10.3f} {area_of_slice:.4f}")

print("-" * 60)
print(f"Total Trapezoidal Approximation: {total_area:.4f}")
```
